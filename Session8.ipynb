```python
#Task 1. <p>Open 'auto_fares.ipynb' and create a list trip_distances of 12 distances between 0.8 and 18.5 km. Use a `for` loop with `enumerate(trip_distances, start=1)` to print 'Trip n: X km -> Rs Y' using your day-time fare rule from the earlier session. Constraint: Do not define any function. Hint: If you missed the earlier session, use a flat Rs 14 per km.</p>

trip_distances = [0.8, 1.2, 1.5, 2, 3.5, 4, 5.8, 6, 7.5, 10, 12.4, 18.5]

for n, distance_km in enumerate(trip_distances, start=1):

    if distance_km <= 1.5:
        fare_inr = 30
    else:
        fare_inr = 30 + (distance_km - 1.5) * 14

    print(f"Trip {n}: {distance_km} km -> Rs {fare_inr:.2f}")
```

```python
#Task 2. <p>Add a `while` loop that adds fares from your trips one by one, using an index variable, and stops as soon as earnings reach Rs 1,500 or the trips run out. Print how many trips were needed and the final earnings.</p>

trip_distances = [0.8, 1.2, 1.5, 2, 3.5, 4, 5.8, 6, 7.5, 10, 12.4, 18.5]

index = 0
total_earnings = 0

while total_earnings < 1500 and index < len(trip_distances):

    distance_km = trip_distances[index]

    if distance_km <= 1.5:
        fare_inr = 30
    else:
        fare_inr = 30 + (distance_km - 1.5) * 14

    total_earnings = total_earnings + fare_inr

    index = index + 1

print("Trips completed:", index)
print("Final earnings: Rs", round(total_earnings, 2))
```

```python
#Task 3. <p>Edit the list so that 2 trips have distance 0 (cancelled) and one trip is above 20 km. Loop through the trips, use `continue` to skip cancelled trips, and use `break` at the first trip above 20 km, printing its position and distance.</p>

trip_distances = [0.8, 0, 1.5, 2, 3.5, 0, 5.8, 6, 7.5, 10, 12.4, 22.5]

for position, distance_km in enumerate(trip_distances, start=1):

    if distance_km == 0:
        continue

    if distance_km > 20:
        print("Trip position:", position)
        print("Trip distance:", distance_km, "km")
        break
```

```python
#Task 4. <p>Finish with a loop that counts Short (under 3 km), Medium (3 to 8 km) and Long (above 8 km) trips in three counter variables and adds the total km and total fare. Print a 3-row summary table plus the totals, then run Restart & Run All.</p>

short_count = 0
medium_count = 0
long_count = 0

total_km = 0
total_fare = 0

for distance_km in trip_distances:

    if distance_km == 0:
        continue

    total_km = total_km + distance_km

    if distance_km <= 1.5:
        fare_inr = 30
    else:
        fare_inr = 30 + (distance_km - 1.5) * 14

    total_fare = total_fare + fare_inr

    if distance_km < 3:
        short_count = short_count + 1
    elif distance_km <= 8:
        medium_count = medium_count + 1
    else:
        long_count = long_count + 1

print("Trip Summary")
print("-------------------------")
print("Category       Trips")
print("-------------------------")
print("Short          ", short_count)
print("Medium         ", medium_count)
print("Long           ", long_count)
print("-------------------------")

# Print overall totals
print("Total KM:", round(total_km, 2))
print("Total Fare: Rs.", round(total_fare, 2))
```
