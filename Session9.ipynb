```python
#Task 1. <p>Create 'coaching_marks.ipynb' for a fictional coaching class. Define `calculate_percentage(marks_obtained, max_marks)` that returns the percentage rounded to 2 decimals, and call it for 3 students. Hint: Use `return`, not `print()`, so the value can be reused later.</p>

def calculate_percentage(marks_obtained, max_marks):
    percentage = (marks_obtained / max_marks) * 100
    return round(percentage, 2)

student1 = calculate_percentage(425, 500)
student2 = calculate_percentage(380, 500)
student3 = calculate_percentage(465, 500)

print("Student 1 Percentage:", student1, "%")
print("Student 2 Percentage:", student2, "%")
print("Student 3 Percentage:", student3, "%")
```

```python
#Task 2. <p>Define `assign_grade(percentage)` returning 'A+' for 90 and above, 'A' from 80, 'B' from 70, 'C' from 60, 'D' from 50 and 'F' below that. Test it on 6 values that include the boundaries 90, 80 and 49.99 and print each result.</p>

test_percentages = [95, 90, 80, 75, 60, 49.99]

for percentage in test_percentages:
    grade = assign_grade(percentage)
    print("Percentage:", percentage, "-> Grade:", grade)
    
def assign_grade(percentage):
    if percentage >= 90:
        return "A+"
    elif percentage >= 80:
        return "A"
    elif percentage >= 70:
        return "B"
    elif percentage >= 60:
        return "C"
    elif percentage >= 50:
        return "D"
    else:
        return "F"    
```

```python
#Task 3. <p>Define `calculate_fee(base_fee, months=6, discount_pct=0)` that returns the total fee after discount, rounded to 2 decimals. Call it once with positional arguments, once with keyword arguments, once using only `base_fee`, and once with a 10 percent discount, printing all four totals.</p>

def calculate_fee(base_fee, months=6, discount_pct=0):
    total_fee = base_fee * months
    discount = total_fee * (discount_pct / 100)
    final_fee = total_fee - discount

    return round(final_fee, 2)

fee1 = calculate_fee(2000, 6, 5)

fee2 = calculate_fee(base_fee=2000, months=3, discount_pct=10)

fee3 = calculate_fee(2000)

fee4 = calculate_fee(2000, discount_pct=10)

print("Positional arguments:", fee1)
print("Keyword arguments:", fee2)
print("Only base fee:", fee3)
print("10% discount:", fee4)
```

```python
#Task 4. <p>Finish by defining `build_report_line(name, marks, max_marks=100)` that uses your two earlier functions and returns a string like 'Name | 78.0% | B'. Store 6 students as a list of tuples, loop through them printing each line, and print the class average percentage. Run Restart & Run All.</p>

def build_report_line(name, marks, max_marks=100):
    percentage = calculate_percentage(marks, max_marks)
    grade = assign_grade(percentage)

    return f"{name} | {percentage}% | {grade}"

students = [
    ("Rahul", 78),
    ("Priya", 95),
    ("Amit", 85),
    ("Neha", 67),
    ("Karan", 49),
    ("Sneha", 90)
]

total_percentage = 0

for name, marks in students:
    report = build_report_line(name, marks)
    print(report)

    percentage = calculate_percentage(marks, 100)
    total_percentage = total_percentage + percentage
    
class_average = total_percentage / len(students)

print("Class Average:", round(class_average, 2), "%")    
```
