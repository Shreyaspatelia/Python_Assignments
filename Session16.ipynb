```python
#Task 2. <p>Create three lists of 6 members: names, plans (Monthly, Quarterly or Yearly) and fees in rupees. Use `zip()` with `enumerate(start=1)` to print lines like '1. Name - Plan - Rs Fee', then build a dictionary of name to fee with `dict(zip(...))`.</p>

names = [
    "Rahul",
    "Priya",
    "Amit",
    "Neha",
    "Karan",
    "Sneha"
]

plans = [
    "Monthly",
    "Quarterly",
    "Yearly",
    "Monthly",
    "Yearly",
    "Quarterly"
]

fees = [
    1500,
    4000,
    12000,
    1500,
    12000,
    4000
]


print("GYM MEMBERSHIP DETAILS")
print("----------------------")

for number, (name, plan, fee) in enumerate(
    zip(names, plans, fees), start=1
):
    print(f"{number}. {name} - {plan} - Rs {fee}")


member_fees = dict(zip(names, fees))

print("\nMember Fee Dictionary:")
print(member_fees)
```

```python
#Task 3. <p>Write a generator function `daily_checkins(days)` that uses `yield` to return a check-in count of `40 + (day * 7) % 25` for each day from 1 to `days`. Print the first 3 values with `next()`, then loop the rest of a 14-day run, and print the total.</p>

def daily_checkins(days):
    for day in range(1, days + 1):
        checkins = 40 + (day * 7) % 25
        yield checkins



checkin_generator = daily_checkins(14)

total_checkins = 0



print("First 3 days:")

for day in range(1, 4):
    value = next(checkin_generator)
    print(f"Day {day}: {value} check-ins")
    total_checkins += value



print("\nRemaining days:")

for day in range(4, 15):
    value = next(checkin_generator)
    print(f"Day {day}: {value} check-ins")
    total_checkins += value


print("\nTotal check-ins for 14 days:", total_checkins)
```

```python
#Task 4. <p>Finish by writing a generator `renewal_alerts(members)` that yields an alert string only for members whose days_left is 7 or less, from a list of 8 member dictionaries you type. Print the generator object, loop through it to print alerts, and print how many were yielded. Run Restart & Run All.</p>

members = [
    {"name": "Rahul", "days_left": 12},
    {"name": "Priya", "days_left": 5},
    {"name": "Amit", "days_left": 3},
    {"name": "Neha", "days_left": 20},
    {"name": "Karan", "days_left": 7},
    {"name": "Sneha", "days_left": 2},
    {"name": "Rohan", "days_left": 15},
    {"name": "Meera", "days_left": 6}
]



def renewal_alerts(members):
    for member in members:
        if member["days_left"] <= 7:
            yield f"Renewal alert: {member['name']} has {member['days_left']} days left!"



alerts = renewal_alerts(members)

print("Generator object:")
print(alerts)



alert_count = 0

print("\nRenewal Alerts:")

for alert in alerts:
    print(alert)
    alert_count += 1



print("\nTotal alerts generated:", alert_count)
```
