```python
#Task 1. <p>Open 'upi_files.ipynb' and define `safe_amount(text)` that converts a string to float inside `try` and returns None on `ValueError`. Test it on "250", "1,200", "abc", "" and "75.5" and print each result. Hint: If you missed the earlier session, start a new notebook with this function only.</p>

def safe_amount(text):
    try:
        amount = float(text)
        return amount

    except ValueError:
        return None


test_values = ["250", "1,200", "abc", "", "75.5"]


for value in test_values:
    result = safe_amount(value)

    print("Input:", repr(value), "| Result:", result)
```

```python
#Task 2. <p>Write a function that opens a file path inside `try`, with `except FileNotFoundError` printing a clear message, an `else` block printing the row count read from 'upi_payments.csv', and a `finally` block printing 'file check done'. Call it with the correct name and with a misspelled one.</p>

import csv

def check_file(file_path):

    try:
        with open(file_path, "r", newline="", encoding="utf-8") as file:
            reader = csv.DictReader(file)
            rows = list(reader)

    except FileNotFoundError:
        print("Error: File not found -", file_path)

    else:
        print("File opened successfully!")
        print("Total rows:", len(rows))

    finally:
        print("file check done")


print("First call:")
check_file("upi_payments.csv")


print("\nSecond call:")
check_file("upi_payment.csv")
```

```python
#Task 3. <p>Create the exception `class InvalidUpiId(Exception): pass` and a function `validate_upi_id(upi_id)` that raises it when the text is empty, has no '@', or has more than one '@'. Test 5 invented IDs such as 'chaishop@upi' and print VALID or the error message. Hint: Catch it with `except InvalidUpiId as e`.</p>

class InvalidUpiId(Exception):
    pass


def validate_upi_id(upi_id):

    if upi_id == "":
        raise InvalidUpiId("UPI ID cannot be empty")

    elif "@" not in upi_id:
        raise InvalidUpiId("UPI ID must contain @")

    elif upi_id.count("@") > 1:
        raise InvalidUpiId("UPI ID cannot contain more than one @")

    else:
        return True


upi_ids = [
    "chaishop@upi",
    "rahul123@okaxis",
    "priyapatel",
    "",
    "amit@upi@bank"
]


for upi_id in upi_ids:

    try:
        validate_upi_id(upi_id)
        print(upi_id, ":", "VALID")

    except InvalidUpiId as e:
        print(upi_id, ":", e)
```

```python
#Task 4. <p>Finish by defining `success_rate(success, total)` that handles `ZeroDivisionError`, and a function `process_payments(rows)` that goes through 10 rows (3 of them bad), counts good and bad rows, writes the bad rows to 'bad_rows.txt' and prints a summary. Run Restart & Run All.</p>

def success_rate(success, total):
    try:
        rate = (success / total) * 100
        return round(rate, 2)

    except ZeroDivisionError:
        print("Cannot calculate success rate: total is zero.")
        return 0
    
    

payment_rows = [
    "TXN001,01/11/2026,Rahul Sharma,500,SUCCESS",
    "TXN002,02/11/2026,Priya Patel,1200,SUCCESS",
    "TXN003,03/11/2026,Amit Shah,350,FAILED",
    "TXN004,04/11/2026,Neha Joshi,abc,SUCCESS",
    "TXN005,05/11/2026,Karan Mehta,1500,SUCCESS",
    "TXN006,06/11/2026,Pooja Desai,250,FAILED",
    "TXN007,07/11/2026,Rohan Dave,SUCCESS",
    "TXN008,08/11/2026,Sneha Shah,650,SUCCESS",
    "TXN009,09/11/2026,Vishal Parmar,900,PENDING",
    "TXN010,10/11/2026,Anjali Modi,1100,SUCCESS"
]


def process_payments(rows):

    good_count = 0
    bad_count = 0
    success_count = 0

    with open("bad_rows.txt", "w", encoding="utf-8") as file:

        for row in rows:

            try:
                # Split the transaction into columns
                fields = row.split(",")

                # Check whether all 5 fields exist
                if len(fields) != 5:
                    raise ValueError("Incorrect number of fields")

                txn_id, date, payee, amount, status = fields

                # Check whether amount is numeric
                amount = float(amount)

                # Check whether status is valid
                if status not in ["SUCCESS", "FAILED"]:
                    raise ValueError("Invalid transaction status")

                # Count valid records
                good_count += 1

                if status == "SUCCESS":
                    success_count += 1

            except ValueError as e:

                bad_count += 1

                # Save incorrect records with their error messages
                file.write(row + " | Error: " + str(e) + "\n")

    rate = success_rate(success_count, good_count)

    print("\nUPI PAYMENT PROCESSING SUMMARY")
    print("--------------------------------")
    print("Total rows:", len(rows))
    print("Good rows:", good_count)
    print("Bad rows:", bad_count)
    print("Successful payments:", success_count)
    print("Success rate:", rate, "%")
    print("Bad records saved in bad_rows.txt")


process_payments(payment_rows)


print("\nTesting zero total:")
print("Success rate:", success_rate(0, 0), "%")
```
