```python
#Task 1. <p>Create 'society_recursion.ipynb' for a fictional housing society. Write a recursive function `sum_dues(dues_list)` that totals 12 monthly maintenance dues you type, with a base case for an empty list and `dues_list[1:]` in the recursive call. Print its result next to `sum(dues_list)` for comparison.</p>

dues_list = [
    1500, 1500, 1600, 1600,
    1700, 1700, 1800, 1800,
    1800, 1900, 1900, 2000
]

def sum_dues(dues_list):
    if len(dues_list) == 0:
        return 0
    else:
        return dues_list[0] + sum_dues(dues_list[1:])

recursive_total = sum_dues(dues_list)
builtin_total = sum(dues_list)

print("Total using recursion:", recursive_total)
print("Total using sum():", builtin_total)
```

```python
#Task 2. <p>Write a recursive function `total_nested(block)` that totals dues stored in a nested list: 3 floors, each holding 4 flat dues. Use `type(item) is list` to decide when to recurse, print the grand total, and check it equals the sum of all 12 values.</p>

society_dues = [
    [1500, 1600, 1550, 1700],
    [1800, 1750, 1600, 1650],
    [1900, 1850, 1700, 1800]
]

def total_nested(block):
    total = 0

    for item in block:
        if type(item) is list:
            total = total + total_nested(item)
        else:
            total = total + item

    return total

grand_total = total_nested(society_dues)

all_dues = sum(sum(floor) for floor in society_dues)

print("Grand Total using recursion:", grand_total)
print("Grand Total using sum():", all_dues)
print("Results match:", grand_total == all_dues)
```

```python
# Task 3. <p>Write a recursive function `late_fee_amount(amount, months, depth=0)` that grows the due by 2 percent per month and prints one line per call with indentation based on depth. Call it with Rs 2,000 for 4 months, then compare the answer with a `for` loop version. Hint: A very large `months` will hit the recursion limit.</p>

def late_fee_amount(amount, months, depth=0):
    print(" " * depth, "Months remaining:", months,
          "| Amount: Rs.", round(amount, 2))

    if months == 0:
        return amount

    new_amount = amount * 1.02

    return late_fee_amount(new_amount, months - 1, depth + 1)


print("Recursive calculation:")

recursive_result = late_fee_amount(2000, 4)

print("\nFinal amount using recursion: Rs.", round(recursive_result, 2))


amount = 2000

for month in range(4):
    amount = amount * 1.02

print("Final amount using for loop: Rs.", round(amount, 2))
print("Results match:", round(recursive_result, 2) == round(amount, 2))
```

```python
#Task 4. <p>Create a global variable maintenance_total = 0 and a function `add_due(amount)` that updates it using the `global` keyword. Add a second function that creates a local variable with the same name, print the global before and after both calls, and keep the UnboundLocalError case as a commented line. Run Restart & Run All.</p>

maintenance_total = 0

def add_due(amount):
    global maintenance_total
    maintenance_total = maintenance_total + amount
    print("Inside add_due:", maintenance_total)


def local_due(amount):
    maintenance_total = amount
    print("Inside local_due:", maintenance_total)


print("Global before calls:", maintenance_total)

add_due(2000)
print("Global after add_due:", maintenance_total)

local_due(5000)
print("Global after local_due:", maintenance_total)

```
