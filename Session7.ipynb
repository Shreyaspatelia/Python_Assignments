```python
# Task 2. <p>Add an if/elif/else ladder that sets trip_label from distance_km: 'Short' below 3, 'Medium' from 3 up to 8, 'Long' above 8 up to 25, and 'Outstation' above 25. Run it for 0.8, 3, 8 and 26 km and keep the four printed results. Hint: Check the boundary values 3, 8 and 25 carefully with `<` and `<=`.</p>

distance_km = 22

if distance_km < 3:
    trip_label = "Short"

elif distance_km <= 8:
    trip_label = "Medium"

elif distance_km <= 25:
    trip_label = "Long"

else:
    trip_label = "Outstation"

print("Distance:", distance_km, "km")
print("Trip Category:", trip_label)
```

```python
# Task 4. <p>Finish by adding an input check that prints 'Invalid trip' when distance_km is 0 or less, and run 5 test trips you choose. For each, print your hand-calculated expected fare, the program's fare using `round(fare_inr, 2)`, and PASS or FAIL. Submit after Restart & Run All.</p>
 
 
  
# Five test trips
test_trips = [
    (0.8, 14, False, 30),
    (1.5, 22, False, 37.5),
    (6, 23, True, 136.25),
    (4, 14, True, 65),
    (0, 2, True, None)
]

for distance_km, hour_24, has_luggage, expected_fare in test_trips:

    if distance_km <= 0:
        print("Invalid trip")
        program_fare = None

    else:
        # Calculate base fare
        if distance_km <= 1.5:
            fare_inr = 30
        else:
            fare_inr = 30 + (distance_km - 1.5) * 14

        # Apply night surcharge
        if hour_24 >= 22 or hour_24 < 5:
            fare_inr = fare_inr + (fare_inr * 0.25)

            # Apply luggage charge
            if has_luggage == True:
                fare_inr = fare_inr + 20

        program_fare = round(fare_inr, 2)

    # Compare expected and actual results
    if program_fare == expected_fare:
        status = "PASS"
    else:
        status = "FAIL"

    print("Distance:", distance_km, "km")
    print("Hour:", hour_24)
    print("Luggage:", has_luggage)
    print("Expected Fare:", expected_fare)
    print("Program Fare:", program_fare)
    print("Result:", status)
    print("--------------------------")
```
