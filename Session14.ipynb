```python
#Task 1. <p>Create 'upi_files.ipynb' for fictional UPI-style payments. With `open()` in 'w' mode, write 'upi_log.txt' with 10 lines in the format txn_id,date,payee,amount_inr,status (dates as DD/MM/YYYY, invented payees, status SUCCESS or FAILED). Append 2 more lines in 'a' mode, read it back with `with open(...)` and print the line count. Hint: Add '\n' to every line and pass `encoding='utf-8'`.</p>

# Step 1: Create 10 fictional UPI transactions
transactions = [
    "TXN001,01/11/2026,Rahul Sharma,500,SUCCESS\n",
    "TXN002,02/11/2026,Priya Patel,1200,SUCCESS\n",
    "TXN003,03/11/2026,Amit Shah,350,FAILED\n",
    "TXN004,04/11/2026,Neha Joshi,800,SUCCESS\n",
    "TXN005,05/11/2026,Karan Mehta,1500,SUCCESS\n",
    "TXN006,06/11/2026,Pooja Desai,250,FAILED\n",
    "TXN007,07/11/2026,Rohan Dave,2000,SUCCESS\n",
    "TXN008,08/11/2026,Sneha Shah,650,SUCCESS\n",
    "TXN009,09/11/2026,Vishal Parmar,900,FAILED\n",
    "TXN010,10/11/2026,Anjali Modi,1100,SUCCESS\n"
]

# Step 2: Write 10 transactions using write mode
with open("upi_log.txt", "w", encoding="utf-8") as file:
    file.writelines(transactions)
    
print("10 transactions written successfully!")


# Step 3: Append 2 more transactions
new_transactions = [
    "TXN011,11/11/2026,Manish Patel,750,SUCCESS\n",
    "TXN012,12/11/2026,Divya Shah,450,SUCCESS\n"
]

with open("upi_log.txt", "a", encoding="utf-8") as file:
    file.writelines(new_transactions)
    
print("2 transactions appended successfully!")


# Step 4: Read the file
with open("upi_log.txt", "r", encoding="utf-8") as file:
    content = file.readlines()

print("\nUPI Transaction Log:")
for line in content:
    print(line.strip())

# Step 5: Print total line count
print("\nTotal number of transactions:", len(content))
```

```python
#Task 2. <p>Create 'upi_payments.csv' with a header and 12 rows using `csv.writer`, then read it with `csv.DictReader` and print the total of amount_inr for SUCCESS rows, converting the amounts with `float()`. Hint: Open the file with `newline=''` when writing CSV.</p>

import csv

# Step 1: Create 12 fictional UPI transactions
transactions = [
    ["TXN001", "01/11/2026", "Rahul Sharma", 500, "SUCCESS"],
    ["TXN002", "02/11/2026", "Priya Patel", 1200, "SUCCESS"],
    ["TXN003", "03/11/2026", "Amit Shah", 350, "FAILED"],
    ["TXN004", "04/11/2026", "Neha Joshi", 800, "SUCCESS"],
    ["TXN005", "05/11/2026", "Karan Mehta", 1500, "SUCCESS"],
    ["TXN006", "06/11/2026", "Pooja Desai", 250, "FAILED"],
    ["TXN007", "07/11/2026", "Rohan Dave", 2000, "SUCCESS"],
    ["TXN008", "08/11/2026", "Sneha Shah", 650, "SUCCESS"],
    ["TXN009", "09/11/2026", "Vishal Parmar", 900, "FAILED"],
    ["TXN010", "10/11/2026", "Anjali Modi", 1100, "SUCCESS"],
    ["TXN011", "11/11/2026", "Manish Patel", 750, "SUCCESS"],
    ["TXN012", "12/11/2026", "Divya Shah", 450, "SUCCESS"]
]

# Step 2: Define CSV column headers
headers = ["txn_id", "date", "payee", "amount_inr", "status"]


# Step 3: Write data into CSV file
with open("upi_payments.csv", "w", newline="", encoding="utf-8") as file:
    writer = csv.writer(file)

    writer.writerow(headers)
    writer.writerows(transactions)

print("CSV file created successfully!")


# Step 4: Read CSV using DictReader
successful_total = 0

with open("upi_payments.csv", "r", newline="", encoding="utf-8") as file:
    reader = csv.DictReader(file)

    for row in reader:
        if row["status"] == "SUCCESS":
            successful_total += float(row["amount_inr"])

# Step 5: Print successful payment total
print("Total successful payments: Rs.", successful_total)
```

```python
# Task 3. <p>Build the dictionary payee_summary with each payee mapped to the total of its SUCCESS amounts, save it as 'payee_summary.json' using `json.dump(..., indent=2)`, load it back with `json.load()` and print the payee with the highest total.</p>

import csv
import json

# Step 1: Create an empty dictionary
payee_summary = {}

# Step 2: Read the CSV file
with open("upi_payments.csv", "r", newline="", encoding="utf-8") as file:
    reader = csv.DictReader(file)

    for row in reader:

        # Consider only successful transactions
        if row["status"] == "SUCCESS":

            payee = row["payee"]
            amount = float(row["amount_inr"])

            # Add amount to existing payee or initialize it
            if payee in payee_summary:
                payee_summary[payee] += amount
            else:
                payee_summary[payee] = amount


# Step 3: Save dictionary into JSON file
with open("payee_summary.json", "w", encoding="utf-8") as file:
    json.dump(payee_summary, file, indent=2)

print("Payee summary saved successfully!")


# Step 4: Load the JSON file
with open("payee_summary.json", "r", encoding="utf-8") as file:
    loaded_summary = json.load(file)


# Step 5: Find the payee with the highest total
highest_payee = max(loaded_summary, key=loaded_summary.get)
highest_amount = loaded_summary[highest_payee]

print("\nPayee Summary:")
print(loaded_summary)

print("\nHighest Payment Payee:", highest_payee)
print("Highest Successful Total: Rs.", highest_amount)
```

```python
#Task 4. <p>Finish with `pathlib`: create the folder 'upi_output' using `Path('upi_output').mkdir(exist_ok=True)`, move the three output files into it by re-saving them there, check each with `.exists()`, print each file's size from `.stat().st_size`, and list the CSV files with `glob('*.csv')`. Run Restart & Run All.</p>

from pathlib import Path

# Step 1: Create the output folder
output_folder = Path("upi_output")
output_folder.mkdir(exist_ok=True)

print("Output folder created successfully!")


# Step 2: Define the three files
file_names = [
    "upi_log.txt",
    "upi_payments.csv",
    "payee_summary.json"
]


# Step 3: Re-save all three files inside upi_output
for file_name in file_names:

    source_file = Path(file_name)
    destination_file = output_folder / file_name

    content = source_file.read_text(encoding="utf-8")
    destination_file.write_text(content, encoding="utf-8")

    print(file_name, "saved inside upi_output")


# Step 4: Check whether each file exists
print("\nFile Existence Check:")

for file_name in file_names:

    file_path = output_folder / file_name

    print(file_name, ":", file_path.exists())


# Step 5: Print the size of each file
print("\nFile Sizes:")

for file_name in file_names:

    file_path = output_folder / file_name
    file_size = file_path.stat().st_size

    print(file_name, ":", file_size, "bytes")


# Step 6: List all CSV files
print("\nCSV Files:")

csv_files = list(output_folder.glob("*.csv"))

for csv_file in csv_files:
    print(csv_file.name)
```
