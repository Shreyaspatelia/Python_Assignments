```python
def price_per_gb(price_inr, data_gb):
    return price_inr / data_gb



def validity_label(days):
    if 28 <= days <= 31:
        return "Monthly"
    elif 84 <= days <= 90:
        return "Quarterly"
    else:
        return "Other"



plans = [
    {"name": "Plan A", "price": 299, "data": 28, "days": 28},
    {"name": "Plan B", "price": 399, "data": 40, "days": 30},
    {"name": "Plan C", "price": 799, "data": 84, "days": 84},
    {"name": "Plan D", "price": 199, "data": 15, "days": 24}
]



print("RECHARGE PLAN ANALYSIS")
print("----------------------")

for plan in plans:
    cost = price_per_gb(plan["price"], plan["data"])
    validity = validity_label(plan["days"])

    print(f"\nPlan: {plan['name']}")
    print(f"Price: Rs {plan['price']}")
    print(f"Data: {plan['data']} GB")
    print(f"Price per GB: Rs {cost:.2f}")
    print(f"Validity: {validity}")
```

```python
#Task 2. <p>In 'recharge_analysis.ipynb' use the `statistics` module on 8 monthly data-usage values in GB you type: print `mean()`, `median()` and `stdev()` rounded to 2 decimals. Also use `math.ceil()` to print how many 28-day recharges cover 100 days.</p>

import statistics
import math


data_usage = [18, 22, 25, 30, 20, 28, 35, 24]


average_usage = statistics.mean(data_usage)
middle_usage = statistics.median(data_usage)
usage_stdev = statistics.stdev(data_usage)



print("MONTHLY DATA USAGE ANALYSIS")
print("---------------------------")

print("Data Usage:", data_usage)
print(f"Mean: {average_usage:.2f} GB")
print(f"Median: {middle_usage:.2f} GB")
print(f"Standard Deviation: {usage_stdev:.2f} GB")



total_days = 100
recharge_validity = 28

recharges_needed = math.ceil(total_days / recharge_validity)

print("\nRECHARGE VALIDITY ANALYSIS")
print("--------------------------")
print("Total Days:", total_days)
print("Recharge Validity:", recharge_validity, "days")
print("Recharges Needed:", recharges_needed)
```

```python
#Task 3. <p>Use `datetime.strptime` with format '%d/%m/%Y' to read the recharge date '05/11/2025', add each of 5 plan validities with `timedelta(days=...)`, and print the expiry dates in DD/MM/YYYY using `strftime`. Create the folder 'recharge_out' with `os.makedirs('recharge_out', exist_ok=True)` and save the results as 'recharge_out/expiry_dates.txt'.</p>

from datetime import datetime, timedelta
import os


recharge_date = datetime.strptime("05/11/2025", "%d/%m/%Y")


plans = [
    {"name": "Plan A", "validity": 28},
    {"name": "Plan B", "validity": 30},
    {"name": "Plan C", "validity": 56},
    {"name": "Plan D", "validity": 84},
    {"name": "Plan E", "validity": 90}
]



os.makedirs("recharge_out", exist_ok=True)


results = []

print("RECHARGE EXPIRY DATE ANALYSIS")
print("-----------------------------")

for plan in plans:

    expiry_date = recharge_date + timedelta(days=plan["validity"])

    formatted_date = expiry_date.strftime("%d/%m/%Y")

    result = f"{plan['name']} - {plan['validity']} days - Expiry Date: {formatted_date}"

    print(result)

    results.append(result)



file_path = "recharge_out/expiry_dates.txt"

with open(file_path, "w") as file:
    file.write("RECHARGE EXPIRY DATE REPORT\n")
    file.write("---------------------------\n")

    for result in results:
        file.write(result + "\n")



print("\nResults saved successfully!")
print("File location:", file_path)
```
