```python
#Task 1. <p>Open 'coaching_marks.ipynb' and create the list students with 10 dictionaries holding name and marks out of 150. Use `map()` with a lambda to build percentages rounded to 1 decimal and print `list()` of the result. Hint: `map()` returns an iterator, so wrap it in `list()` to see the values. If you missed the earlier session, start a new notebook with this list.</p>

students = [
    {"name": "Rahul", "marks": 145},
    {"name": "Priya", "marks": 132},
    {"name": "Amit", "marks": 120},
    {"name": "Neha", "marks": 110},
    {"name": "Karan", "marks": 98},
    {"name": "Sneha", "marks": 87},
    {"name": "Rohan", "marks": 150},
    {"name": "Pooja", "marks": 75},
    {"name": "Vishal", "marks": 140},
    {"name": "Anjali", "marks": 105}
]

percentages = map(
    lambda student: round((student["marks"] / 150) * 100, 1),
    students
)

print(list(percentages))
```

```python
#Task 2. <p>Use `filter()` with a lambda to keep students above 75 percent and print the count and names. Then use `sorted()` with `key=lambda s: s['marks']` and `reverse=True` to print the top 3 students by marks.</p>

above_75 = list(filter(
    lambda s: (s["marks"] / 150) * 100 > 75,
    students
))

print("Number of students above 75%:", len(above_75))

print("Students above 75%:")
for student in above_75:
    print(student["name"])
    
sorted_students = sorted(
    students,
    key=lambda s: s["marks"],
    reverse=True
)

print("Top 3 Students:")

for student in sorted_students[:3]:
    print(student["name"], "-", student["marks"], "marks")    
    
```

```python
#Task 3. <p>Import `reduce` from `functools` and use it with lambdas on a list of 10 fee payments you type: first to get the total fees collected, then to find the highest single payment without using `max()`. Print both results.</p>
 
from functools import reduce

fee_payments = [5000, 4500, 6000, 5500, 4000,
                7000, 3500, 6500, 4800, 7500]

total_fees = reduce(
    lambda a, b: a + b,
    fee_payments
)

print("Total fees collected: Rs.", total_fees)

highest_payment = reduce(
    lambda a, b: a if a > b else b,
    fee_payments
)

print("Highest single payment: Rs.", highest_payment)
```

```python
#Task 4. <p>Finish by defining `apply_to_all(func, items)` that returns a list, and use it with three different lambdas: names in title case, marks plus 5 grace marks capped at 150 using `min()`, and percentages converted to grades by calling `assign_grade`. Print all three outputs and run Restart & Run All.</p>

def calculate_percentage(marks_obtained, max_marks):
    percentage = (marks_obtained / max_marks) * 100
    return round(percentage, 2)

def assign_grade(percentage):
    if percentage >= 90:
        return "A+"
    elif percentage >= 80:
        return "A"
    elif percentage >= 70:
        return "B"
    elif percentage >= 60:
        return "C"
    elif percentage >= 50:
        return "D"
    else:
        return "F"


students = [
    {"name": "Rahul", "marks": 145},
    {"name": "Priya", "marks": 132},
    {"name": "Amit", "marks": 120},
    {"name": "Neha", "marks": 110},
    {"name": "Karan", "marks": 98},
    {"name": "Sneha", "marks": 87},
    {"name": "Rohan", "marks": 150},
    {"name": "Pooja", "marks": 75},
    {"name": "Vishal", "marks": 140},
    {"name": "Anjali", "marks": 105}
]


def apply_to_all(func, items):
    return list(map(func, items))

names = apply_to_all(
    lambda s: s["name"].title(),
    students
)
print("Student Names:")
print(names)


grace_marks = apply_to_all(
    lambda s: min(s["marks"] + 5, 150),
    students
)
print("\nMarks After Grace:")
print(grace_marks)


grades = apply_to_all(
    lambda s: assign_grade(
        calculate_percentage(s["marks"], 150)
    ),
    students
)

print("\nStudent Grades:")
print(grades)
```
