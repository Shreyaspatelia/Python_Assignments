```python
# Task 2. Write a function update_playlist_price(playlist, new_price) that updates the price of a given playlist in the playlist_prices dictionary. Test it by updating the price of any one playlist and printing the updated dictionary.

playlist_prices = {
    "Top Hits": 199,
    "Bollywood Hits": 149,
    "Punjabi Beats": 99,
    "Workout Music": 129,
    "Romantic Songs": 179
}

def update_playlist_price(playlist, new_price):
    playlist_prices[playlist] = new_price

update_playlist_price("Top Hits", 299)

print(playlist_prices)
```

```python
# Task 4. Given two sets: set1 contains the names of restaurants you have ordered from on Zomato, and set2 contains the names of restaurants you have ordered from on Swiggy, find and print the union and intersection of these sets.<br><br><em><strong>Hint:</strong> Use the union() and intersection() methods of Python sets.</em>

set1 = {"Dominos", "McDonalds", "Subway"}
set2 = {"McDonalds", "Subway", "Pizza Hut"}

union_set = set1.union(set2)
intersection_set = set1.intersection(set2)

print("Union:", union_set)
print("Intersection:", intersection_set)
```
